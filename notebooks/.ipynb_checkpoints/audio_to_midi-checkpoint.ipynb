# vi.ve — Audio → MIDI 🎚️➡️🎹

Drop full songs into **`trell_music_life/songs_to_load/`**, run this notebook, and each song becomes a **MIDI pack in your naming convention** — extracted into the separate `midi_university_auto/` library so machine transcriptions never mix with your hand-played ground truth.

**In plain english, what happens to each song:**
1. A neural net **un-bakes the cake** — the finished mix is split back into 4 ingredient stems (vocals / drums / bass / other)
2. The song's **key and BPM are detected** from the audio itself
3. Each melodic stem goes through a **pitch-tracking model** that reads the spectrogram and writes down every note it hears (pitch, start, length, loudness)
4. The drum stem gets **sliced at every hit**, and each hit is sorted into kick / snare / hihat by where its energy lives in the frequency spectrum
5. Everything is written as `{title} key {Key} bpm {bpm}/` with `{title} - part.mid` files, and the song is added to **the log** (`data/songs_loaded.csv`) with quality columns for your grading

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

INK, INK2, MUTED, SURFACE = "#0b0b0b", "#52514e", "#898781", "#fcfcfb"
BLUE, AQUA, YELLOW, VIOLET = "#2a78d6", "#1baf7a", "#eda100", "#4a3aa7"

fig, ax = plt.subplots(figsize=(11, 4.6), facecolor=SURFACE)
ax.set_xlim(0, 100); ax.set_ylim(0, 46); ax.axis("off")

def box(x, y, w, h, text, color, tc="white", fs=9.5):
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.6",
                                fc=color, ec="none"))
    ax.text(x + w/2, y + h/2, text, ha="center", va="center",
            fontsize=fs, color=tc, fontweight="bold")

def arrow(x1, y1, x2, y2):
    ax.add_patch(FancyArrowPatch((x1, y1), (x2, y2), arrowstyle="-|>",
                                 mutation_scale=14, color=MUTED, lw=1.4))

box(1, 19, 12, 8, "song file\n(mp3 / m4a / wav)", INK2)
box(19, 19, 13, 8, "DEMUCS\nstem separation", BLUE)
box(19, 4, 13, 7, "key + BPM\ndetection", VIOLET)
box(40, 34, 14, 7, "vocals stem", AQUA, fs=9)
box(40, 24.5, 14, 7, "other stem", AQUA, fs=9)
box(40, 15, 14, 7, "bass stem", AQUA, fs=9)
box(40, 4, 14, 7, "drums stem", AQUA, fs=9)
box(61, 24.5, 15, 16.5, "BASIC PITCH\nspectrogram →\nnote events", YELLOW, tc="#0b0b0b")
box(61, 4, 15, 7, "onset slice +\nkick/snare/hat", YELLOW, tc="#0b0b0b")
box(84, 14, 14, 12, "MIDI PACK\n+ song log\n(your convention)", "#008300")

arrow(13, 23, 19, 23)
arrow(25, 19, 25, 11.4)
for y in (37.5, 28, 18.5):
    arrow(32, 23, 40, y)
arrow(32, 21, 40, 7.5)
for y in (37.5, 28, 18.5):
    arrow(54, y, 61, 32 if y == 37.5 else (30 if y == 28 else 27))
arrow(54, 7.5, 61, 7.5)
arrow(76, 30, 84, 22)
arrow(76, 7.5, 84, 16)
ax.text(25.5, 1.2, "key + bpm stamp every MIDI file", fontsize=8, color=MUTED, style="italic")

ax.set_title("songs_to_load  →  stems  →  notes  →  midi_university_auto",
             loc="left", fontsize=12, color=INK, pad=14)
plt.tight_layout(); plt.show()
```

## 4 · What to listen for when grading

- **Bassline** — usually the most accurate stem (monophonic, clear pitch). Check octaves: pitch trackers sometimes jump an octave on subby 808s.
- **Melody (other stem)** — this is piano/synth/guitar mashed together; expect the most extra notes. Dense chords are the hardest case for any transcription model.
- **Vocal melody** — good for the *contour* of the hook; vibrato and slides can smear into note clusters.
- **Drums** — kick detection is strong; snare-vs-hat classification is the roughest edge (crossover energy). Timing is raw (not quantized) — that's intentional: the microtiming *is* data.
- **Key/BPM** — spot-check against your ear. Trap BPMs get folded into the 50–100 range (a "140" song logs as 70 — same grid, half-time frame).

## 5 · What happens after grading

1. You mark quality + notes in the log → we tune `CONFIG` and re-run (`force=True`)
2. Once quality is trustworthy: a **new Postgres table** stores extracted-song info so vi.ve can pull it when building new styles
3. Extracted packs can then flow into the same analysis pipeline as your hand-played packs (`parse_packs.py` → metrics → style maps) — with a provenance flag so the system always knows machine-extracted from hand-played
